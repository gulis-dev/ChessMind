# Nierównomierne shardy – skrót

---


## 1. Problem
Shardy (pliki) mają skrajnie różne lub jednorodne rozkłady faz (np. jeden shard ≈100% OPEN). Powstało to przez brak globalnego tasowania przed cięciem albo sekwencyjne generowanie danych.

---

## 2. Skutki w treningu
- Bias na początku epoki (model „widzi” tylko jeden typ pozycji, niestabilne gradienty).
- W DDP różne workery uczą się z innych faz → rozjazd statystyk i wolniejsza konwergencja.
- Batch/feature normy oparte na wąskim wycinku danych → gorsza generalizacja.
- Early stopping po części epoki daje mylące wyniki (ocena tylko jednej fazy).

---


## 3. Minimalne rozwiązanie
Globalnie potasować wszystkie rekordy przed podziałem lub jednocześnie mieszać wiele shardów w DataLoaderze (np. losowy wybór wierszy z puli N shardów). Jeśli zasoby małe: jednorazowe scalenie → shuffle → równe nowe shardy.
