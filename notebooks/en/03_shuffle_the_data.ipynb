# Uneven Shards – Summary

---

## 1. Problem
Shards (files) have highly skewed or nearly single‑phase distributions (e.g. one shard ≈100% OPEN). This comes from missing a global shuffle before slicing or from sequential data generation.

---

## 2. Training Impact
- Early‑epoch bias (the model “sees” only one type of position → unstable gradients).
- In DDP, different workers learn from different phases → diverging statistics, slower convergence.
- Batch/feature norms built on a narrow slice → worse generalization.
- Early stopping after a partial epoch gives misleading metrics (evaluation reflects only one phase).

---

## 3. Minimal Mitigation
Globally shuffle all records before re‑sharding, or mix multiple shards concurrently in the DataLoader (e.g. random row sampling from a pool of N shards). If resources are tight: one‑off concatenate → shuffle → split into equal new shards.
